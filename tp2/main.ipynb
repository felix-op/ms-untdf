## Vector DOMINIO

```python
import numpy as np
import matplotlib.pyplot as plt

vector_t = np.arange(0, 100.1, 0.1)

def escalon_base(vector):
    return np.where(vector >= 0, 1, 0)

def dibujar(titulo, funcion):
    plt.plot(vector_t, funcion)
    plt.xlabel("t")
    plt.ylabel("f(t)")
    plt.title(titulo)
    plt.grid()
    plt.show()
```

1. **Escalón**\
    $x(t) = A\,\mu(t-t_0)$

```python
AMPLITUD = 10
T0 = 20

def escalon(A, vector, t0):
    return A * escalon_base(vector - t0)

resultado_escalon = escalon(AMPLITUD, vector_t,  T0)
dibujar("Función escalón", resultado_escalon)
```

2. **Seno**\
    $x(t) = A\sin(2\pi f_0t+\theta)$

```python
AMPLITUD = 5
FRECUENCIA = 0.07
FASE = 2

def seno(A, f0, vector, fase):
    return A * np.sin(2 * np.pi * f0 * vector + fase)

resultado_seno = seno(AMPLITUD, FRECUENCIA, vector_t, FASE)
dibujar("Función seno", resultado_seno)
```

3. **Exponencial**\
    $x(t) = A e^{\alpha(t-t_0)}$

```python
AMPLITUD = 5
ALPHA = 0.07
T0 = 2

def exponencial(A, alpha, vector, t0):
    return A * np.exp(alpha * (vector - t0))

resultado_exponencial = exponencial(AMPLITUD, ALPHA, vector_t, T0)
dibujar("Función exponencial", resultado_exponencial)
```

4. **Señal amortiguada**\
    $x(t) = A\sin(2\pi f_0t+\theta)e^{\alpha t}$

```python
AMPLITUD = 5
FRECUENCIA = 0.07
FASE = 2
ALPHA = 0.07
T0 = 2

def amortiguada(A, f0, vector, fase, alpha):
    return seno(A, f0, vector, fase) * exponencial(1, alpha, vector, 0)

resultado_amortiguada = amortiguada(AMPLITUD, FRECUENCIA, vector_t, FASE, ALPHA)
dibujar("Función de señal amortiguada", resultado_amortiguada)
```

5. **Pulso**\
    $p(t) = A[\mu(t-t_0)-\mu(t-t_0-D)]$


```python
AMPLITUD = 5
T0 = 40
D = 20

def pulso(A, vector, t0, D):
    return A * (escalon_base(vector - t0) - escalon_base(vector - t0 - D))

resultado_pulso = pulso(AMPLITUD, vector_t, T0, D)
dibujar("Función Pulso", resultado_pulso)
```
